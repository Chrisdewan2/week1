```python
labels = []

for s in df1[4]:
    if s == 'Iris-setosa':
        labels.append(0)
    elif s == 'Iris-versicolor':
        labels.append(1)
    else:
        labels.append(2)

df1['Species_label'] = labels

print(df1['Species_label'].unique())
df1
```

```python
cols = [0, 1, 2, 3]        # only the numeric columns, column 4 is text

print("Mean")
print(df1[cols].mean())
print("\nMedian")
print(df1[cols].median())
print("\nMin")
print(df1[cols].min())
print("\nMax")
print(df1[cols].max())
```

```python
import numpy as np
print(np.mean(df1[0]))
```

```python
g = df1.groupby('Species_label')[cols]

print("MEAN")
print(g.mean())
print("\nVAR")
print(g.var())
print("\nMIN")
print(g.min())
print("\nMAX")
print(g.max())
```

```python
import matplotlib.pyplot as plt

colors = ['red', 'green', 'blue']
names  = ['Iris-setosa', 'Iris-versicolor', 'Iris-virginica']

plt.figure(figsize=(7,5))

for i in range(3):
    sub = df1[df1['Species_label'] == i]
    plt.scatter(sub[2], sub[3], c=colors[i], label=names[i])


```
